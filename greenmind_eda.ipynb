This notebook conducts an in-depth **Exploratory Data Analysis (EDA)** on edge telemetry logs to:
1. Quantify **material stream composition** and landfill diversion efficacy.
2. Model **avoided greenhouse gas emissions ($CO_2e$)** resulting from diversion.
3. Audit **machine learning prediction certainty & confidence calibration**.
4. Deliver **prescriptive business recommendations** for municipal resource recovery.

### 3. Descriptive Statistics & Environmental KPIs

```python
# Aggregated KPI computations
total_scans = len(df)
bio_scans = (df['category'] == 'biodegradable').sum()
non_scans = (df['category'] == 'non-biodegradable').sum()
diversion_rate = (bio_scans / total_scans) * 100
total_mass_kg = df['weight_kg'].sum()
bio_mass_diverted_kg = df[df['category'] == 'biodegradable']['weight_kg'].sum()
total_co2_averted_kg = df['co2e_avoided_kg'].sum()
avg_confidence = df['confidence_pct'].mean()

print("=" * 50)
print("           EXECUTIVE SUSTAINABILITY REPORT        ")
print("=" * 50)
print(f"• Total Scans Processed:           {total_scans}")
print(f"• Clean Diversion Rate:            {diversion_rate:.1f}%")
print(f"• Total Mass Throughput:           {total_mass_kg:.3f} kg")
print(f"• Landfill Mass Diverted:          {bio_mass_diverted_kg:.3f} kg")
print(f"• Net CO2e Emissions Avoided:      {total_co2_averted_kg:.3f} kg CO2e")
print(f"• Mean Model Confidence:           {avg_confidence:.2f}%")
print("=" * 50)
```

### 4. Material Stream Breakdown & Composition Analysis

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# 1. Material Category Frequency
material_counts = df['material'].value_counts()
axes[0].barh(material_counts.index, material_counts.values, color='#2d7a2d', edgecolor='#1b4d1b')
axes[0].set_title('Waste Volume by Material Category', fontsize=12, fontweight='bold')
axes[0].set_xlabel('Scan Count')

# 2. Landfill Diversion Stream Ratio
category_counts = df['category'].value_counts()
axes[1].pie(category_counts, labels=['Biodegradable', 'Non-Bio Landfill Risk'], autopct='%1.1f%%', 
           colors=['#2d7a2d', '#c0392b'], explode=[0.05, 0], startangle=140,
           wedgeprops={'edgecolor': 'white', 'linewidth': 2})
axes[1].set_title('Stream Diversion Ratio', fontsize=12, fontweight='bold')

plt.tight_layout()
plt.show()
```

### 5. Machine Learning Calibration & Confidence Distribution

```python
plt.figure(figsize=(10, 4))
sns.boxplot(data=df, x='category', y='confidence_pct', palette=['#c0392b', '#2d7a2d'])
sns.stripplot(data=df, x='category', y='confidence_pct', color='black', alpha=0.6, jitter=0.2, size=7)
plt.title('MobileNet Model Confidence Calibration across Stream Categories', fontsize=12, fontweight='bold')
plt.xlabel('Waste Category')
plt.ylabel('Confidence Score (%)')
plt.axhline(85, color='orange', linestyle='--', label='High Confidence Benchmark (85%)')
plt.legend(loc='lower left')
plt.show()
```

### 6. Cumulative Carbon Avoidance Trajectory

```python
# Compute cumulative running carbon avoidance
df_chrono = df.sort_values('timestamp').copy()
df_chrono['cum_co2e'] = df_chrono['co2e_avoided_kg'].cumsum()

plt.figure(figsize=(11, 4))
plt.plot(df_chrono['scan_id'], df_chrono['cum_co2e'], marker='o', color='#2d7a2d', linewidth=2.5, label='Cumulative CO2e Avoided')
plt.fill_between(df_chrono['scan_id'], df_chrono['cum_co2e'], color='#2d7a2d', alpha=0.15)
plt.title('Cumulative Environmental Carbon Mitigation Trajectory', fontsize=12, fontweight='bold')
plt.xlabel('Telemetry Scan ID')
plt.ylabel('Avoided CO2e (kg)')
plt.legend()
plt.grid(True, linestyle=':', alpha=0.6)
plt.show()
```
